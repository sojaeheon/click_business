```python
import pandas as pd
```

```python
# CSV 파일 불러오기
df = pd.read_csv("../raw/kbs_news_gold_2020_2025.csv", encoding="utf-8-sig") # 금
df = pd.read_csv("../raw/kbs_news_interest_2020_2025.csv", encoding="utf-8-sig") # 금리
df
```

```python
# 컬럼 확인
df.columns
```

```python
# 데이터 정보 확인
print("전체 데이터 개수:", len(df))
print(df.info())
```

```python
# 제목 기준으로 중복 제거
df = df.drop_duplicates(subset='Title', keep='first').reset_index(drop=True)
print("전체 데이터 개수:", len(df))
```

```python
# 날짜 기준 정렬 및 년도별 기사 개수
df['Date'] = pd.to_datetime(df['Date'].astype(str), format='%Y%m%d', errors='coerce')
df['Year'] = df['Date'].dt.year

df = df.sort_values('Date').reset_index(drop=True)

news_count_per_year = df.groupby('Year').size().reset_index(name='ArticleCount')
print(news_count_per_year)

df = df.drop(columns=["Year"], errors="ignore")
```

```python
# CSV 저장
df.to_csv("../raw/kbs_news_gold_2020_2025.csv", index=False, encoding="utf-8-sig") # 금
df.to_csv("../raw/kbs_news_interest_2020_2025.csv", index=False, encoding="utf-8-sig") # 금리
print("CSV 저장 완료")
```
